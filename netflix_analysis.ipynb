```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")

plt.savefig("movies_vs_tvshows.png")
plt.show()
```

```python
country_count = df['country'].value_counts().head(10)

country_count.plot(kind='bar', figsize=(10,5))

plt.title("Top 10 Countries with Most Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.show()
```

```python
df['release_year'].value_counts().sort_index().tail(20).plot(
    kind='line',
    figsize=(10,5),
    marker='o'
)

plt.title("Netflix Content Released Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
rating_count = df['rating'].value_counts().head(10)

rating_count.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Top 10 Netflix Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

plt.savefig("netflix_ratings.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
genre_count = df['listed_in'].value_counts().head(10)

genre_count.plot(
    kind='barh',
    figsize=(10,6)
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.savefig("top_genres.png", dpi=300, bbox_inches="tight")

plt.show()
```

# Key Findings

1. Netflix contains more Movies than TV Shows.
2. United States has the highest number of Netflix titles.
3. India is the second-largest content producer.
4. Netflix content increased significantly after 2015.
5. TV-MA and TV-14 are the most common ratings.
6. Drama and International content are among the most popular genres.
